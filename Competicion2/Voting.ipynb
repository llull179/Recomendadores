```python
import pandas as pd

lora = pd.read_csv('data/submissions/submission_lora_bert_base_multilingual_uncased_sentiment.csv')
dl = pd.read_csv('data/submissions/DL_rounded.csv')
dl2 = pd.read_csv('data/submissions/DL_rounded2.csv')
transformers = pd.read_csv('data/submissions/submission_transformer-bert-base-sentiment.csv')
```

```python
lista_votadores = [lora,dl,dl2,transformers]

voting = pd.DataFrame()
first = True
for i in range(0,len(lista_votadores)):
    dataset = lista_votadores[i]
    if i == 0:
        voting['review_id'] = dataset['review_id']
        first = False
    voting['pred'+str(i)] = dataset['stars']
```

```python
voting_columns = voting.drop(columns=['review_id']).columns.tolist()
```

```python
voting['stars'] = voting[voting_columns].mean(axis=1, skipna=True).round(0).astype(int)
voting[['review_id','stars']].to_csv('./data/submissions/mean_ensemble.csv', index=False)
```

```python
voting['stars2'] =voting[voting_columns].mode(axis=1)[0]
```

```python
voting_ensemble = voting[['review_id','stars2']].copy()
voting_ensemble['stars'] = voting_ensemble['stars2'].astype(int)
voting_ensemble = voting_ensemble.drop(columns=['stars2'])
voting_ensemble.to_csv('./data/submissions/voting_ensemble.csv', index=False)
voting_ensemble.head()
```
